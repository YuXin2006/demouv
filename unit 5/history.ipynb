```python
from langchain_core.documents import Document

documents=[
    Document(
        page_content="珠穆朗玛峰，简称珠峰，是喜马拉雅山脉的主峰，也是地球上海拔最高的山峰。",
        metadata={"source":1},
    ),
    Document(
        page_content="藏语： “珠穆朗玛”(Qomolangma)意为“大地之母”或“女神”",
        metadata={"source":2},
    ),
    Document(
        page_content="珠峰顶部的气候极端恶劣。由于海拔极高，顶峰的大气含氧量仅为海平面的 1/3 左右。",
        metadata={"source":3},
    ),
    Document(
        page_content="首次成功登顶： 1953年5月29日,新西兰登山家埃德蒙·希拉里（Edmund Hillary）和尼泊尔向导丹增·诺盖（Tenzing Norgay）从南坡首次登顶。",
        metadata={"source":4},
    ),
    Document(
        page_content="珠穆朗玛峰的形成是由于 印度板块 与 欧亚板块 的持续碰撞。",
        metadata={"source":5},
    )
]

documents
```

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()
os.environ["SILICONFLOW_API_KEY"]=os.getenv("SILICONFLOW_API_KEY")

llm = ChatOpenAI(
    model_name='Pro/zai-org/GLM-4.7', 
    openai_api_key=os.environ["SILICONFLOW_API_KEY"],
    openai_api_base='https://api.siliconflow.cn/v1', # 关键：指向国内加速网关
    streaming=True
)
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(
    model="BAAI/bge-m3", # 硅基流动上的模型 ID
    openai_api_key=os.environ["SILICONFLOW_API_KEY"],
    openai_api_base="https://api.siliconflow.cn/v1" # 同样指向这个网关
)
```

```python
from langchain_community.vectorstores import FAISS
db=FAISS.from_documents(documents, embedding=embeddings)
db
```

```python
await db.asimilarity_search("气候")
```

```python
db.similarity_search_with_score("气候")
```

```python

from langchain_core.runnables import RunnableLambda
# 这里优化了一下similarity_search_with_score方法 ，将其封装成一个RunnableLambda对象 支持 .batch()：自动开启多线程/异步处理。支持lcel语法：可以直接连接到 prompt 或 llm 后面。
retriever=RunnableLambda( db.similarity_search_with_score).bind(k=1)
retriever.batch(["气候","含氧量"])
```

```python
retriever = db.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 1}
)

results = retriever.batch(["气候", "含氧量"])
results
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough

messages = """
使用提供的文本回答我的问题
{question}

context:
{context}
"""

prompt = ChatPromptTemplate.from_messages([("human", messages)])
rag_chain={"context":retriever,"question":RunnablePassthrough()}|prompt|llm
response=rag_chain.invoke("告诉我一些关于珠峰的气候的信息")
response.content
```
