```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from dotenv import load_dotenv


load_dotenv()
#from langchain.agents import create_react_agent, AgentExecutor
#from langchain import hub

# 1. 定义工具 (Tool)
@tool
def magic_calculator(expression: str) -> str:
    """这是一个魔法计算器，可以计算复杂的数学表达式。输入应为字符串，如 '23 * 45'。"""
    try:
        # 实际执行的 Python 代码
        return str(eval(expression))
    except Exception as e:
        return f"计算错误: {str(e)}"

tools = [magic_calculator]
```

```python
llm = ChatOpenAI(
    model_name='Pro/zai-org/GLM-4.7', 
    openai_api_key=os.environ["SILICONFLOW_API_KEY"],
    openai_api_base='https://api.siliconflow.cn/v1',
    temperature=0 # Agent 必须设置低随机性，否则容易逻辑混乱
)
```

```python
#import langchain_hub as hub
#prompt = hub.pull("hwchase17/react")
from langchain_core.prompts import PromptTemplate

# 这就是 hwchase17/react 的内容（中文微调版，更适合 GLM-4）
template = """尽力回答以下问题。你可以使用以下工具：

{tools}

请按照以下格式进行回答：

Question: 输入的问题
Thought: 你应该总是思考接下来该做什么
Action: 采取行动的工具名称（应为 [{tool_names}] 之一）
Action Input: 采取行动的工具输入参数
Observation: 工具输出的结果
... (这个 Thought/Action/Action Input/Observation 过程可以重复多次)
Thought: 我现在知道最终答案了
Final Answer: 对原始问题的最终回答

开始！

Question: {input}
Thought: {agent_scratchpad}"""

prompt = PromptTemplate.from_template(template)
print(prompt)
```

```python
from langchain.agents import create_agent
agent=create_agent(
    model=llm,
    tools=tools,
)
```

```python
from langchain.agents import AgentExecutor,create_openai_function_agent
agent_executor = AgentExecutor(
    agent=agent, 
    tools=tools, 
    verbose=True,
    handle_parsing_errors=True
)
```
