```python
import logging

# 基础配置
logging.basicConfig(
    level=logging.DEBUG, # 设置级别: DEBUG, INFO, WARNING, ERROR, CRITICAL
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    filemode='w',
    filename='app.log'#指定日志数据存放好的文件
)

logging.debug("this is a debug msg")

```
