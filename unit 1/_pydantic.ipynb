```python
from pydantic import BaseModel

class User(BaseModel):
    id: int
    name: str
    age: int = 18  # 默认值
    
u = User(id=1, name="张三")
print(u)    
```

```python
from typing import Optional
class User(BaseModel):
    id:int
    name:str
    age:int
    salary:Optional[float]=None#默认为none
    is_active:Optional[bool]=True#默认为true
u=User(id=1,name='李四',age=18)
print(u)   
```
