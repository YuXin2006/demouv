```python
df1=pd.DataFrame({'name':['张三','李四','王五'],'age':[18,20,22]})
df2=pd.DataFrame({'name':['李四','王二','麻子'],'age':[20,22,28]})
pd.merge(df1,df2,on='name',how='inner')#内合并(交集)

```

```python
pd.merge(df1,df2,on='name',how='outer')#外合并(并集)
```

```python
pd.merge(df1,df2,on='name',how='left')#左合并
pd.merge(df1,df2,on='name',how='right')#右合并
```
